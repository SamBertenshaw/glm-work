```python
import pandas as pd
```

```python
claims = pd.read_csv("C:\\Users\\sambe\\.vscode\\random code\\glm work\\insurance_claims.csv")

filtered_df = claims[["policy_number","policy_annual_premium","total_claim_amount", "age"]].round(2)
display(filtered_df)
```

```python
grouped = (filtered_df.groupby("age")
            .agg({"policy_annual_premium": "mean", "total_claim_amount": "mean"})
            .reset_index()
            .round(2))

display(grouped)
```
